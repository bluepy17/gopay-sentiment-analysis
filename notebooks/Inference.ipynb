```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Load LSTM model
model = load_model('lstm_sentiment_model.keras')

# Load tokenizer
with open('tokenizer_lstm.pkl', 'rb') as handle:
    tokenizer = pickle.load(handle)

# Label encoder
le = LabelEncoder()
le.classes_ = np.array(['positive', 'negative', 'netral'])
# Fungsi inference
def predict_sentiment_lstm(text_list, max_len=100):
    """
    text_list : list of strings
    max_len : panjang sequence padding (harus sama dengan saat training)
    """
    # Tokenisasi + padding
    sequences = tokenizer.texts_to_sequences(text_list)
    padded = tf.keras.preprocessing.sequence.pad_sequences(sequences, maxlen=max_len)

    # Prediksi
    preds = model.predict(padded)
    pred_labels = np.argmax(preds, axis=1)

    # Mapping ke label asli
    pred_labels = le.inverse_transform(pred_labels)
    return pred_labels
```

```python

texts = [
    "Aplikasi ini bagus sekaliii!",
    "Aplikasinya lemot banget, banyak fitur gak jelas"
]

predictions = predict_sentiment_lstm(texts)
for t, p in zip(texts, predictions):
    print(f"Teks: {t}\nPrediksi Sentimen: {p}\n")
```
