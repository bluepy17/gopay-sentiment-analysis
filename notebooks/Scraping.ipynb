*   **Topik :** Analisis Sentimen Review Aplikasi Gopay di Play Store
*   **Platform Sumber Data :** Google Play Store



```python
!pip install google-play-scraper

# Mengakses review dan informasi aplikasi dari Google Play Store.
from google_play_scraper import app, reviews, reviews_all, Sort

# Pandas untuk manipulasi dan analisis data
import pandas as pd
pd.options.mode.chained_assignment = None  # Menonaktifkan peringatan chaining

# Menyimpan review ke dalam file CSV
import csv
```

```python
# Mengambil semua review dari aplikasi
scrapreview,continuation_token = reviews(
    'com.gojek.gopay',        # ID aplikasi
    lang='id',                # Bahasa review pengguna
    country='id',             # Negara
    sort=Sort.MOST_RELEVANT,  # Urutan review
    count=15000               # Jumlah maksimum ulasan yang ingin diambil
)

print(f"Total review yang berhasil diambil: {len(scrapreview)}")
```

```python
with open('data_raw.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['Review'])  # Menulis header kolom
    for review in scrapreview:
        writer.writerow([review['content']])  # Menulis konten review ke dalam file CSV

# Membuat DataFrame dari hasil scrapreview
df_raw = pd.DataFrame(scrapreview)

# Menyimpan DataFrame ke CSV
df_raw.to_csv('data_raw.csv', index=False)

# Menghitung jumlah baris dan kolom dalam DataFrame
jumlah_ulasan, jumlah_kolom = df_raw.shape

print(f"Jumlah Ulasan : {df_raw.shape[0]}")
print(f"Jumlah Kolom  : {df_raw.shape[1]}")

df_raw.head()
```
